# Ejercicio 3: Compresor de texto


```python
def comprimir_texto(texto):
   letras = []
   caracter_actual = texto[0]
   contador= 1

   for i in range(1, len(texto)):
      if texto[i] == caracter_actual:
         contador += 1
      else:
         letras.append(f"{caracter_actual}{contador}")
         caracter_actual = texto[i]
         contador = 1

   letras.append(f"{caracter_actual}{contador}")
   return "".join(letras)
```

```python
def pedir_texto():
   while True:
      texto = input("Ingrese un texto (no vacío) para ser comprimido: ")

      if texto.strip():
         comprimido = comprimir_texto(texto)
         print(f"Texto comprimido: {comprimido}")
         break
      else:
         print("Ingrese un texto no vacío, por favor.")
```

```python
texto_prueba = "aaabbcaaa"
texto_comprimido = comprimir_texto(texto_prueba)
print(texto_comprimido)

pedir_texto()
```

# Ejercicio 4: Validación de contraseña

```python
def validar_contraseña(contrasena):
    requisitos_incumplidos = []  

    if len(contrasena) < 10:
        requisitos_incumplidos.append("La contraseña debe tener al menos 10 caracteres.")
    if not any(letra.isupper() for letra in contrasena ):
        requisitos_incumplidos.append("La contraseña debe tener al menos una letra mayúscula.")
    if not any(letra.islower() for letra in contrasena):
        requisitos_incumplidos.append("La contraseña debe tener al menos una letra minúscula.")
    if not any(letra.isdigit() for letra in contrasena):
        requisitos_incumplidos.append("La contraseña debe tener al menos un dígito.")
    # if not any(letra in ",;.:-_{[¨*]}"¡?'¿=)(/&%$#"!°|") for letra in contrasena):
        requisitos_incumplidos.append("La contraseña debe tener al menos un carácter especial.")
    
    for i in range(len(contrasena) - 2):
        if contrasena[i] == contrasena[i + 1] == contrasena[i + 2]:
            requisitos_incumplidos.append("La contraseña no debe contener tres caracteres iguales consecutivos.")
            break # Basta con una coincidencia para incumplir el requisito
    
    if not requisitos_incumplidos:
        print("Cuenta creada correctamente")
    else:
        print("La contraseña no cumple con los siguientes requisitos:")
        for requisito in requisitos_incumplidos:
            print(f"- {requisito}")
        print("Validacion fallida.")
```

```python
def registro_nuevo_usuario():

   nombre_usuario = input("Ingrese su nombre de usuario: ")
   mensaje_contrasena = f""" 
Bienvenido {nombre_usuario}, para crear tu cuenta, por favor ingresa una contraseña que cumpla con los siguientes requisitos:
- Al menos 10 caracteres de longitud.
- Al menos una letra mayúscula, una minúscula y un caracter especial.
- No debe contener tres caracteres iguales consecutivos.
"""
   contrasena = input(mensaje_contrasena)
   validar_contraseña(contrasena)
```

```python
validar_contraseña("Aa1!Aa1!Aa1!")
validar_contraseña("luisroman123")
```

```python
registro_nuevo_usuario()
```
